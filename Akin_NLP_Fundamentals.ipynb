```python
import pandas as pd
from sklearn.feature_extraction.text import CountVectorizer, TfidfVectorizer

texts = [
    "I love machine learning",
    "machine learning is powerful",
    "I love NLP",
    "NLP is fun"
]
```

#Exercise1


```python
tokens= "I love machine learning".lower().split()
print (tokens)
```

#Exercise 2

```python
vectorizer = CountVectorizer()
X_bow = vectorizer.fit_transform(texts)
print("Matrix shape:", None if X_bow is None else X_bow.shape)
```

#Exercise 3

```python
vocab = vectorizer.get_feature_names_out()
print("Vocabulary:", None if vocab is None else vocab)
print(vectorizer.vocabulary_)
```

#Exercise 4

```python
bow_df = pd.DataFrame(X_bow.toarray(), columns=vocab)
print(bow_df)
```

#Exercise 5

```python
tfidf_vectorizer = TfidfVectorizer()
X_tfidf = tfidf_vectorizer.fit_transform(texts)
tfidf_df = pd.DataFrame(X_tfidf.toarray(), columns=tfidf_vectorizer.get_feature_names_out())
tfidf_df
```

#Relection

1. What does Bag-of-Words represent?
Bag-of-Words turns a sentence into a list of numbers that shows how many times each word appears. It doesn’t care about the order of the words. It only counts them. So “dog bites man” and “man bites dog” look exactly the same to it. In my notebook, each of the 4 sentences became a row of 7 numbers, one for each word.

2. What is the main difference between Bag-of-Words and TF-IDF?
Bag-of-Words just counts words, so every word counts the same. TF-IDF gives each word a score for how important it is. A word that shows up in lots of sentences, like “learning”, gets a lower score because it doesn’t tell us much. A word that shows up in only one sentence, like “powerful”, gets a higher score because it’s more special to that sentence. I could see this in my TF-IDF table.

3. Why do we need to convert text into numbers?
A machine learning model can only work with numbers, because it does maths like adding and multiplying. It can’t do that with words. So we change the text into numbers first, and then the model can look for patterns and make predictions.
