```python
# Exercise 2
def sigmoid(x):
    return 1 / (1 + np.exp(-x))

print(sigmoid(0))
print(sigmoid(2))
```

```python
# Exercise 3
def relu(x):
    return max(0, x)

print(relu(-3))
print(relu(5))
```

```python
# Exercise 4
x = np.linspace(-5, 5, 100)

y = sigmoid(x)

plt.plot(x, y)
plt.title("Sigmoid Function")
plt.xlabel("Input")
plt.ylabel("Output")
plt.show()
```

```python
# Exercise 5 - hidden layer
X = np.array([1.5, 2.0])

W_hidden = np.array([
    [0.3, -0.2],
    [0.4, 0.1]
])

b_hidden = np.array([0.1, -0.1])

hidden = sigmoid(np.dot(X, W_hidden) + b_hidden)

print("hidden =", hidden)
```

```python
# Exercise 5 - output
W_output = np.array([0.6, -0.3])
b_output = 0.05

output = np.dot(hidden, W_output) + b_output

print("Network output:", output)
```

#1. i noticed a weight decides how much an input pushes the answer. A positive weight pushes the answer up, a negative weight pulls it down, and a bigger weight makes that effect stronger.

#2. The bias is a number added to the answer no matter what the inputs are. It shifts the result up or down, so that a neuron can have a starting value even when all the inputs are zero.

#3. Without activation functions, a network can only learn straight-line (linear) patterns. They add a bend to the maths, so the network can learn more complicated patterns, and they squash the numbers into a smaller range, like sigmoid does with 0 to 1.
